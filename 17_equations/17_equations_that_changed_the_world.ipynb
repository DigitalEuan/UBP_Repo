```python
# @title 17 Black-Scholes
import numpy as np
from scipy.integrate import solve_ivp

# #17 Black-Scholes: Simple 1D finite diff proxy for C(t,S) PDE
# Discretize: ∂C/∂t + (1/2)σ²S² ∂²C/∂S² + r S ∂C/∂S - r C = 0
# Use explicit FD on grid, σ=0.2, r=0.05, T=1, S=100, K=100 (call payoff max(S-K,0))
def bs_pde(t, C, sigma=0.2, r=0.05, S_grid=np.linspace(50,150,50)):
    dS = S_grid[1] - S_grid[0]
    dtau = 0.01  # dt proxy
    # FD: C_new = C + dt * [ -0.5 sigma^2 S^2 d2C/dS2 - r S dC/dS + r C ]
    dC_dS = np.gradient(C, dS)
    d2C_dS2 = np.gradient(dC_dS, dS)
    drift = -0.5 * sigma**2 * S_grid**2 * d2C_dS2 - r * S_grid * dC_dS + r * C
    return drift / dtau  # Approx for solve_ivp (reverse time)

S_grid = np.linspace(50,150,50)
C0 = np.maximum(S_grid - 100, 0)  # Initial payoff at t=T
sol = solve_ivp(bs_pde, [0,1], C0, t_eval=[0], args=(), method='RK45')
C_final = sol.y[:,-1]  # At t=0
# BS exact proxy at S=100: ~10.45
bs_exact = 10.45
nrc_i17 = 1 - abs(C_final[25] - bs_exact) / bs_exact  # Midpoint S=100
print(f'17: C_final_mid={C_final[25]:.3f}, Exact={bs_exact}, NRCI={nrc_i17:.3f}')
```
